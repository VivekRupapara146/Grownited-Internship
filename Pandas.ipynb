##Lambda Function

Its a one line function which takes multiple arguments but ony have one expression.

syntax:
```python
func_name = lambda arguments : expression

# examples

square = lambda x : x*x # one argument
add = lambda a,b,c : a+b+c # multiple arguments

# but both has a single expression
```

````python
#Normal func
def square(x):
    return x * x
print(square(6))
# same using lambda
sq = lambda x : x * x
print(sq(5))
````

##Map()

- map() is a built-in pandas function used to apply a function to every item in an iterable (like a list, tuple, or series) and return the results. Iyts same as pandas apply() function

```python
map(function, iterable)
# Example:
```
- Ends when shortest iterable ends

````python
nums = [1, 2, 3, 4]

result = map(lambda x: x * 2, nums)
print(list(result))
````

````python
# using UDF
def square(x):
    return x * x

nums = [1, 2, 3]
print(list(map(square, nums)))
````

````python
# map() can take more than one iterable.
a = [1, 2, 3]
b = [4, 5, 6]

result = map(lambda x, y: x + y, a, b)
print(list(result))
````

````python
#  using for loop
result = []
for x in nums:
  result.append(x*2)
print(result)
````

````python
rsl = list(map(lambda x : x*2, nums))
print(rsl)
````

##filter()

- used to select elements from a iterable based on condition
```
filter(function, iterable)
```
- each element is tested
- returned obj is converted to list

````python
nums = [1, 2, 3, 4, 5]

result = filter(lambda x: x % 2 == 0, nums)
print(list(result))
````

````python
print(list(filter(lambda x : x%2==0, nums)))
````

````python
words = ["cat", "", "dog", "", "bird"]

list(filter(lambda x: x != "", words))
````
